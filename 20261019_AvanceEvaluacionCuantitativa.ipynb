```r
library(tidyverse)
library(readxl)

#Carga datos
datos <- read_excel("ptsd.xlsx", sheet=2)
```

Limpieza de Datos

```r
#Datos nulos
colSums(is.na(datos))
```

```r
#Tipos de datos
glimpse(datos)
```

```r
#Transformar a categóricas
datos<- datos %>%
  mutate(
  a1_Sexo = as.factor(a1_Sexo),
  a8_Años_Estudio_Categoria = factor(a8_Años_Estudio_Categoria, levels= c("Basica", "Media", "Superior", "Analfabeto")),
  PTSD_Diagnostico_UA = as.factor(PTSD_Diagnostico_UA),
  EEV_EC_1_Fallecimiento_familiar = as.factor(EEV_EC_1_Fallecimiento_familiar),
  EEV_EC_2_Atrapado_y_rescatado = as.factor(EEV_EC_2_Atrapado_y_rescatado),
  EEV_EC_3_Lesiones_fisicas_Graves_Hospitalizacion = as.factor(EEV_EC_3_Lesiones_fisicas_Graves_Hospitalizacion),
  EEV_EC_4_Daño_grave_o_Perdida_Vivienda = as.factor(EEV_EC_4_Daño_grave_o_Perdida_Vivienda),
  EEV_EC_Final = as.factor(EEV_EC_Final)
  )
```

```r
#Eliminar duplicados
sum(duplicated(datos))
```

```r
head(datos)
```

Primer Análisis

```r
modelo_parte1 <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final + EEV_TE_Suma_TOTAL,
                     data = datos,
                     family = binomial(link = "logit"))
summary(modelo_parte1)
```

```r
cat("El valor de AIC es:", AIC(modelo_parte1), "\n")
cat("El valor de Odds Ratio es:", exp(coef(modelo_parte1)), "\n")
```

```r
library(tidyverse)
library(readxl)
library(broom)

#Carga datos
datos <- read_excel("ptsd.xlsx", sheet=2)
```

Limpieza de Datos

```r
#Datos nulos
colSums(is.na(datos))
```

```r
#Tipos de datos
glimpse(datos)
```

```r
#Transformar a categóricas
datos<- datos %>%
  mutate(
  a1_Sexo = as.factor(a1_Sexo),
  a8_Años_Estudio_Categoria = factor(a8_Años_Estudio_Categoria, levels= c("Basica", "Media", "Superior", "Analfabeto")),
  PTSD_Diagnostico_UA = as.factor(PTSD_Diagnostico_UA),
  EEV_EC_1_Fallecimiento_familiar = as.factor(EEV_EC_1_Fallecimiento_familiar),
  EEV_EC_2_Atrapado_y_rescatado = as.factor(EEV_EC_2_Atrapado_y_rescatado),
  EEV_EC_3_Lesiones_fisicas_Graves_Hospitalizacion = as.factor(EEV_EC_3_Lesiones_fisicas_Graves_Hospitalizacion),
  EEV_EC_4_Daño_grave_o_Perdida_Vivienda = as.factor(EEV_EC_4_Daño_grave_o_Perdida_Vivienda),
  EEV_EC_Final = as.factor(EEV_EC_Final)
  )
```

```r
#Eliminar duplicados
sum(duplicated(datos))
```

```r
head(datos)
```

Primer Análisis

```r
modelo_parte1 <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final + EEV_TE_Suma_TOTAL,
                     data = datos,
                     family = binomial(link = "logit"))
summary(modelo_parte1)
```

```r
tidy(modelo_parte1, exponentiate = TRUE, conf.int = TRUE )%>%
  mutate_if(is.numeric, round, 5)
```

Segundo Análisis (+ depresión + edad) estratificado por sexo

```r
# Modelo solo para Mujeres
modelo_mujeres <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final +
                      EEV_TE_Suma_TOTAL + f32YE_Depresion_Simple_UA + a2_Edad,
                      data = subset(datos, a1_Sexo == "Femenino"),
                      family = binomial(link = "logit"))
summary(modelo_mujeres)


# Modelo solo para Hombres
modelo_hombres <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final +
                      EEV_TE_Suma_TOTAL + f32YE_Depresion_Simple_UA + a2_Edad,
                      data = subset(datos, a1_Sexo == "Masculino"),
                      family = binomial(link = "logit"))
summary(modelo_hombres)
```

```r

tidy(modelo_mujeres, exponentiate = TRUE, conf.int = TRUE) %>%
  mutate_if(is.numeric, round, 5)
tidy(modelo_hombres, exponentiate = TRUE, conf.int = TRUE)%>%
  mutate_if(is.numeric, round, 5)
```

```r
# Modelo solo para Mujeres (agregado escolaridad)
modelo_mujeres_esc <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final +
                      EEV_TE_Suma_TOTAL + f32YE_Depresion_Simple_UA + a2_Edad +  a8_Años_Estudio_Categoria ,
                      data = subset(datos, a1_Sexo == "Femenino"),
                      family = binomial(link = "logit"))
summary(modelo_mujeres_esc)


# Modelo solo para Hombres (agregado escolaridad)
modelo_hombres_esc <- glm(PTSD_Diagnostico_UA ~ Violencias_infancias + EEV_EC_Final +
                      EEV_TE_Suma_TOTAL + f32YE_Depresion_Simple_UA + a2_Edad +  a8_Años_Estudio_Categoria ,
                      data = subset(datos, a1_Sexo == "Masculino"),
                      family = binomial(link = "logit"))
summary(modelo_hombres_esc)
```
